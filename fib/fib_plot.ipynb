```python
import matplotlib.pyplot as plt
import numpy as np

n_values = [1, 5, 10, 15, 20, 25, 30, 35, 40]

time_recursive = [0, 2, 13, 99, 415, 4587, 86362, 915479, 21291922]
time_iterative = [1, 2, 3, 5, 12, 13, 21, 16, 21]
time_top_down = [5, 9, 4, 10, 12, 12, 4, 5, 19]
time_bottom_up = [1, 4, 4, 5, 4, 6, 4, 5, 4]

plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(12, 8))

ax.plot(n_values, time_iterative, marker='o', linestyle='-', label='Iterative O(N)')
ax.plot(n_values, time_top_down, marker='s', linestyle='-', label='Top-Down DP O(N)')
ax.plot(n_values, time_bottom_up, marker='^', linestyle='-', label='Bottom-Up DP O(N)')
ax.plot(n_values, time_recursive, marker='x', linestyle='--', color='red', label='Recursive O(2^N)')

ax.set_title('Performance Comparison of Fibonacci Algorithms', fontsize=16)
ax.set_xlabel('Number of Terms (N)', fontsize=12)
ax.set_ylabel('Time Taken (microseconds)', fontsize=12)

ax.set_yscale('log')

ax.legend(fontsize=10)
ax.grid(True, which="both", ls="--")

ax.set_xticks(n_values)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

```
