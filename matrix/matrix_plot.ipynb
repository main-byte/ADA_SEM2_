```python
import matplotlib.pyplot as plt
sizes=[16,32,64,128,256,512]
iterative_times=[0.000100,0.00162,0.001317,0.011225,0.078401,0.648881]
recursive_times=[0.000364,0.002832,0.0026803,0.0177581,1.422480,6.773926]
strassen_times=[0.000422,0.002656,0.0021366,0.127811,0.862122,3.376650]
plt.plot(sizes, iterative_times, color= 'yellow', marker='o')
plt.plot(sizes, recursive_times, color= 'blue', marker='o')
plt.plot(sizes, strassen_times, color= 'red', marker='o')
plt.xlabel("Input Size (n)")
plt.ylabel("Time Taken (sec)")
plt.title("Linear Iterative: Time vs Input Size")
plt.grid(True)
plt.show()
```
